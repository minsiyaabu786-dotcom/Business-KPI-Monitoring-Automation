```python
import pandas as pd
df = pd.read_excel("uncleaned dataset.xlsx")
df["Actual Value"] = pd.to_numeric(df["Actual Value"], errors='coerce')
df["Actual Value"] = df["Actual Value"].fillna(df["Actual Value"].median())
```

```python
import pandas as pd
df["Previous Period Value"] = pd.to_numeric(df["Previous Period Value"], errors='coerce')
df["Previous Period Value"] = df["Previous Period Value"].fillna(df["Previous Period Value"].median())
```

```python
print("duplicate rows")
print(df.duplicated().sum())
```

```python
df=df.drop_duplicates()
```

```python
print("unique values")
for col in df.select_dtypes(include="object"):
    print(f"{col}")
    print(df[col].unique())
```

```python
df.columns = df.columns.str.strip()
df["Date"] = df["Date"].str.strip().str.replace('_', ' ', regex=False).str.lower()
_date_map = {
    '19-jun-2024': '2024-06-19',
    '2026/03/17': '2026-03-17',
    '03/20/2025': '2025-03-20',
    '10-oct-2024': '2024-10-10',
    '2025/03/25': '2025-03-25'
}
df["Date"] = df["Date"].replace(_date_map)
print(df["Date"].unique())
```

```python
# Convert stripped date formats securely
def parse_cleansed_date(val):
    val = str(val).strip()
    for fmt in ('%Y%m%d', '%d%m%Y'):
        try:
            return pd.to_datetime(val, format=fmt)
        except ValueError:
            continue
    return pd.to_datetime(val, errors='coerce')

df['Date'] = df['Date'].apply(parse_cleansed_date)
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
```

```python
df["Department"] = df["Department"].str.strip().str.replace('_', ' ', regex=False).str.lower().replace({"hr": "Human Resources", "sales": "Sales", "operations": "Operations", "finance": "Finance", "marketing": "Marketing", "human resources": "Human Resources"})
print(df["Department"].unique())
```

```python
df["KPI_Name"] = df["KPI_Name"].str.strip().str.replace('_', ' ', regex=False).str.lower().str.title()
print(df["KPI_Name"].unique())
```

```python
df["KPI Category"] = df["KPI Category"].str.strip().str.replace('_', ' ', regex=False).str.lower().replace({
    'hr': 'Human Resources',
    'operational': 'Operational',
    'financial': 'Financial',
    'marketing': 'Marketing'})
print(df["KPI Category"].unique())
```

```python
df["KPI Status"] = df["KPI Status"].str.strip().str.replace('_', ' ', regex=False).str.lower().replace({
    'on track': 'On Track',
    'off track': 'Off Track',
    'unknown': 'Unknown'})
print(df["KPI Status"].unique())
```

```python
df["KPI_Name"] = df["KPI_Name"].str.strip().str.lower()
corrections = {'Ord3R Fulfillm3Nt Tim3': 'Order Fulfillment Time', 'Custom3R Acquisition Cost': 'Customer Aquisition Cost', 'Employ33 Churn Rt3': 'Employee Churn Rate',
 'Conv3Rsion Rt3': 'Conversion Rate', 'W3Bsit3 Trffic': 'Website Traffic', 'Syst3M Uptim3 P3Rc3Ntg3': 'System Uptime Percentage',
 'Trining Hours P3R Employ33': 'Training Hours Per Employee', 'Op3Rting Mrgin': 'Operating Margin', 'Employ33 Stisfction Scor3': 'Employee Satisfaction Score',
 'Ontim3 D3Liv3Ry Rt3': 'Ontime Delivery Rate', 'Monthly R3V3Nu3': 'Monthly Revenue', 'L3D G3N3Rtion Count': 'Lead Generation Count'}
df["KPI_Name"] = df["KPI_Name"].replace(corrections)
```

```python
import numpy as np
for col in df.select_dtypes(include="number").columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    print(f"{col} : {count} ouliers")
```

```python
for col in df.select_dtypes(include="number").columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[col] = df[col] = df[col].clip(lower_bound, upper_bound)
    print("Outliers handled successfully")
    print (df)
```
